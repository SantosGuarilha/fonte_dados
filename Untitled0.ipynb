```python
import pandas as pd

url = 'https://raw.githubusercontent.com/ybifoundation/Dataset/main/Salary%20Data.csv'
df = pd.read_csv(url)
display(df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression

# Prepare the data
X = df[['Experience Years']]
y = df['Salary']

# Create and fit the linear regression model
model = LinearRegression()
model.fit(X, y)

# Get the intercept and coefficient
intercept = model.intercept_
coefficient = model.coef_[0]

# Create the regression plot
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Experience Years', y='Salary', data=df, s=100, alpha=0.7)
sns.lineplot(x='Experience Years', y=model.predict(X), data=df, color='red', linewidth=2)

plt.title('Linear Regression: Salary vs. Experience Years')
plt.xlabel('Experience Years')
plt.ylabel('Salary')
plt.grid(True)
plt.show()

print(f"Intercept: {intercept:.2f}")
print(f"Coefficient (Experience Years): {coefficient:.2f}")
```
